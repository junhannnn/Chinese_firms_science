```python
import os
import re
import pandas as pd
```

```python
data = pd.read_csv("../../../proc_output/WOS_listed/CSMAR_subs_name.csv").drop_duplicates(subset=['RalatedParty'])
data
```

```python
out_dir = "../../../dataset/CSMAR_subsidiary/CSMAR_subs_Engname/to_search"

chinese_pattern = re.compile(r'[\u4e00-\u9fff]')

def has_chinese(s: str) -> bool:
    return bool(chinese_pattern.search(s))
```

```python
series = (
    data['RalatedParty']
    .dropna()
    .astype(str)
)

filtered = [s for s in series if has_chinese(s)]
filtered
```

```python
len(filtered)
```

```python
max_len = 19500
chunks = []
current_chunk = ""

for item in filtered:
    # separate entries by newline (change to other separator if needed)
    text_to_add = (item if current_chunk == "" else "\n" + item)
    if len(current_chunk) + len(text_to_add) > max_len:
        # save current chunk and start a new one
        chunks.append(current_chunk)
        current_chunk = item  # start new chunk without leading newline
    else:
        current_chunk += text_to_add

# add last chunk if non-empty
if current_chunk:
    chunks.append(current_chunk)

for idx, chunk in enumerate(chunks):
    out_path = os.path.join(out_dir, f"{idx+1}.txt")  # use 0,1,2,...; change if you prefer 1-based
    with open(out_path, "w", encoding="utf-8") as f:
        f.write(chunk)
```
