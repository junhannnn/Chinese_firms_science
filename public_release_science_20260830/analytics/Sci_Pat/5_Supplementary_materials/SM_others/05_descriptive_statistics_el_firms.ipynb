```python
# Load config
dataset_config = yaml.safe_load(open("config/dataset.yml", encoding="utf-8"))

# Get output folder
figure_output = dataset_config["path_figure"] + 'CN_CN/'
if figure_output is None:
    raise KeyError("Cannot find 'path_figure' in config/dataset.yml")

os.makedirs(figure_output, exist_ok=True)

# ===== Data =====
categories = ["Individuals", "Universities/Institutes", "Gov. Entities", "Firms"]
values = [26, 138, 21, 499]
hatch_patterns = ["///", "xx", "..", "--"]

# ===== Plot =====
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

bars1 = axes[0].bar(
    categories,
    values,
    color=["#ff9999", "#66b3ff", "#99ff99", "#ffcc99"],
    edgecolor="white",
    linewidth=1.2
)
for bar, hatch in zip(bars1, hatch_patterns):
    bar.set_hatch(hatch)

axes[0].set_title(
    "U.S. Entity List: Mainland China-related Entries (Up to June 2024)\nTotal = 684",
    fontsize=12,
    weight="bold"
)
axes[0].set_ylabel("Number of Entries")
axes[0].bar_label(bars1)

firm_categories = ["With Patent Activity", "No Patent Activity"]
firm_values = [293, 206]
x_pos = [0, 0.8]

bars2 = axes[1].bar(
    x_pos,
    firm_values,
    color=["#ffb3e6", "#c2c2f0"],
    edgecolor="white",
    linewidth=1.2,
    width=0.4
)
for bar, hatch in zip(bars2, hatch_patterns[:2]):
    bar.set_hatch(hatch)

axes[1].set_title(
    "Firms Breakdown (499 Mainland China Firms)",
    fontsize=12,
    weight="bold"
)
axes[1].set_ylabel("Number of Firms")
axes[1].bar_label(bars2)

axes[1].set_xticks(x_pos)
axes[1].set_xticklabels(firm_categories)

plt.tight_layout()

# ===== Save to figure_output =====
out_path = os.path.join(figure_output, "background_EL_number.png")
fig.savefig(out_path, dpi=600, bbox_inches="tight")

plt.show()
print("Saved:", out_path)
```
