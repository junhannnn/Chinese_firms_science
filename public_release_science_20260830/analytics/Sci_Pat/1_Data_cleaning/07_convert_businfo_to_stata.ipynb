```python
import pandas as pd
import yaml
```

```python
import os
os.chdir('../../../')
```

```python
dataset_config = yaml.safe_load(open("config/dataset.yml"))
dataset_config
```

```python
businfo = pd.read_stata(dataset_config['path_cnpat'] + "1985-2025/businfo.dta")
businfo
```

## SOE

```python
businfo['soe'] = businfo['工商公司类型'].apply(lambda x: 1 if '国有' in str(x) else 0)
businfo
```

```python
businfo_soe = businfo[['申请号','soe']]
businfo_soe = businfo_soe.rename(columns={"申请号":"apn"})
businfo_soe
```

```python
businfo_soe.to_csv(dataset_config['path_processed'] + "CN_CN/businfo_soe.csv", index=False, encoding='utf-8', sep="\t")
```

## English name and foreign

```python
businfo.rename(columns={"申请号":"apn", "工商别名":"name", "工商英文名":"eng_name", "工商注册地址":"address","工商公司类型":"type","工商成立日期":"found_date", "工商统一社会信用代码": "social_credit_code", "工商注册号":"registration_num", "工商上市代码":"list_code", "工商企业状态":"status"}, inplace=True)
businfo
```

```python
businfo_sim = businfo[['apn','name','eng_name','type']].dropna(subset="type")
businfo_sim
```

```python
# Define the pattern for the regex search
pattern = "港|澳|台|外商|外国|中外"

# Update the 'foreign' column based on the regex pattern
businfo_sim['foreign'] = businfo_sim['type'].astype(str).str.contains(pattern, regex=True)
businfo_sim
```

```python
businfo_sim = businfo_sim.drop(columns='type')
businfo_sim
```

```python
businfo_sim.to_csv(dataset_config['path_processed'] + "CN_CN/businfo_simple.csv", index=False, encoding='utf-8', sep="\t")
```
