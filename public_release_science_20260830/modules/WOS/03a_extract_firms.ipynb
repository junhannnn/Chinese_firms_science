```python
import pandas as pd

# Load the CSV
df = pd.read_csv("02_WOS_unique_orgs_with_type.csv")
df
```

```python
# Filter for US and CN
filtered = df[df["type"].isin(["Company"])]
filtered
```

```python
filtered_CN = filtered[filtered.country == 'CN']
filtered_CN
```

```python
filtered_US = filtered[filtered.country == 'US']
filtered_US
```

```python
filtered_CN.to_parquet('03_WOS_unique_orgs_firms_CN.parquet')
filtered_US.to_parquet('03_WOS_unique_orgs_firms_US.parquet')
```
