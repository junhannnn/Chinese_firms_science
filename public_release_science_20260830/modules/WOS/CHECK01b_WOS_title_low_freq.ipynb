## Import data

```python
df_low_freq = pd.read_csv(dataset_config['path_processed'] + 'WOS/CHECK01_WOS_low_freq.csv')
df_low_freq
```

```python
df_wos_title = pd.read_parquet(dataset_config['path_processed'] + 'WOS/WOS_paper_title.parquet').drop_duplicates()
df_wos_title
```

```python
df_final = df_low_freq.merge(df_wos_title)
df_final
```

```python
df_final_sorted = df_final.sort_values(by='year', ascending=False)
df_final_sorted
```

```python
df_final_sorted = df_final_sorted[~df_final_sorted['organizations'].isin([
    'Magnitude Biosci',
    'Looxid Labs US Inc',
    'Florida Electrophysiol',
    'Allnamics USA'
])]
df_final_sorted
```

```python
df_final_sorted[['organizations', 'total_papers', 'wosid', 'title', 'journal', 'year', 'WOS area']].to_csv(dataset_config['path_processed'] + 'WOS/CHECK01_WOS_title_low_freq.csv', index=False)
```
