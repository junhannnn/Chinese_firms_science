```python
import pandas as pd
import yaml
from pathlib import Path

from pandarallel import pandarallel
pandarallel.initialize(progress_bar=True, nb_workers=16)
```

```python
# Load configuration settings from a YAML file.
# The configuration file typically contains file paths and other parameters.
dataset_config = yaml.safe_load(open("config/dataset.yml"))
dataset_config
```

```python
wos_path = Path(dataset_config['path_wos'] + 'WOS_Articles_20241010/')
```

```python
# Import OpenAlex data. Executive time: 6 min
wos = []
for p in wos_path.iterdir():
    # Extract the year from the filename (assuming filenames are in the format 'year.csv')
    year = int(p.stem)
    
    # Only load files from 2000 to 2022
    if 2010 <= year <= 2023:
        print('Loading', p)
        df_item = pd.read_csv(p, usecols=['wosid', 'researcher_id', 'seq_no', 'organizations', 'reprint']).dropna(subset=['organizations'])
        wos.append(df_item)
```

```python
df_wos = pd.concat(wos)
df_wos
```

```python
df_wos.to_csv(dataset_config['path_processed'] + 'WOS/WOS_2010_2023.csv', index=False)
```
